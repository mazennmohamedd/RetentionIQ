```python
total_charges=pd.to_numeric(df["TotalCharges"], errors="coerce")

print("Invalid TotalCharges values:", total_charges.isnull().sum())

print(df.loc[
    total_charges.isna(),[
        "customerID", "TotalCharges", "tenure", "MonthlyCharges","Churn"
    ]
])
```

```python
df["TotalCharges"]=pd.to_numeric(
    df["TotalCharges"].str.strip(),
    errors="coerce")

df.loc[
    df["TotalCharges"].isna() & (df["tenure"]==0),
    "TotalCharges"
]=0

print("Missing TotalCharges:", df["TotalCharges"].isna().sum())
print("Data type:", df["TotalCharges"].dtype)
print("Customers with tenure = 0:", (df["tenure"] == 0).sum())

df.loc[df["tenure"] == 0, ["tenure", "MonthlyCharges", "TotalCharges", "Churn"]].head(12)
```

```python
churn_summary=(
    df["Churn"]
    .value_counts()
    .rename_axis("Churn")
    .to_frame("Customers")
)

churn_summary["Percentage"] = (
    churn_summary["Customers"] / len(df) * 100
).round(2)

churn_summary
```

```python
assert df["TotalCharges"].notna().all()
assert df["Churn"].isin(["Yes", "No"]).all()
assert not df.duplicated().any()

print("Data inspection and cleaning completed successfully.")
print(f"Final shape: {df.shape}")
```

```python
OUTPUT_PATH = Path("../data/processed/telco_clean.csv")

OUTPUT_PATH.parent.mkdir(parents=True, exist_ok=True)
df.to_csv(OUTPUT_PATH, index=False)

print(f"Cleaned dataset saved to: {OUTPUT_PATH}")
```
